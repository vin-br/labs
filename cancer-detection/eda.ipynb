# Breast Cancer Detection
## 1. Exploratory Data Analysis

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import polars as pl

from common.visualizations import (
    plot_class_distribution,
    plot_feature_distributions,
    plot_scatter_pairs,
)
```

```python

data = pl.read_parquet("data/breast_cancer.parquet")
```

### Dataset Structure

```python
data.head(10)
```

```python
print(f"Shape:   {data.shape}")
print(f"Columns: {data.columns}")
print(data.dtypes)
```

### Cleaning

```python
# Drop empty column, rename columns with spaces
data = (
    data
    .drop([col for col in data.columns if col.startswith("Unnamed")])
)

# Build rename mapping only for remaining columns
rename_map = {col: col.replace(" ", "_") for col in data.columns}
data = data.rename(rename_map)

# Verify no nulls remain
print(data.null_count())
```

### Class Distribution

```python
plot_class_distribution(data).show()
```

### Descriptive Statistics

```python
data.describe()
```

### Feature Distributions by Diagnosis

```python
plot_feature_distributions(data, [
    "radius_mean", "texture_mean", "area_mean", "smoothness_mean",
]).show()
```

### Feature Relationships

```python
plot_scatter_pairs(data, [
    ("radius_mean", "fractal_dimension_mean", "Radius vs Fractal Dimension"),
    ("radius_mean", "area_mean",              "Radius vs Area"),
    ("radius_mean", "texture_mean",           "Radius vs Texture"),
    ("area_mean",   "smoothness_mean",         "Area vs Smoothness"),
], title="Feature Relationships").show()
```

```python
plot_scatter_pairs(data, [
    ("area_mean",       "fractal_dimension_mean", "Area vs Fractal Dimension"),
    ("texture_mean",    "fractal_dimension_mean", "Texture vs Fractal Dimension"),
    ("smoothness_mean", "fractal_dimension_mean", "Smoothness vs Fractal Dimension"),
    ("symmetry_mean",   "fractal_dimension_mean", "Symmetry vs Fractal Dimension"),
], title="Fractal Dimension Comparisons").show()
```

```python
plot_scatter_pairs(data, [
    ("area_mean",    "smoothness_mean", "Area vs Smoothness"),
    ("texture_mean", "smoothness_mean", "Texture vs Smoothness"),
], title="Smoothness Comparisons").show()
```
