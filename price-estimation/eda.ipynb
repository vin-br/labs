# Exploratory Data Analysis

Subject: House Prices Prediction

Goal : Predict the sales price for each house in Ames, Iowa. For each Id in the test set, you must predict the value of the SalePrice variable.

Metric : Root-Mean-Squared-Error (RMSE) between the logarithm of the predicted value and the logarithm of the observed sales price. (Taking logs means that errors in predicting expensive houses and cheap houses will affect the result equally.)

### Dealing with missing values

Before going any further, we will check if there are missing values in this dataframe and we will deal with these missing values.

Some columns seem to have null data. But when we look at the data description file given with the dataset, it seems that some data with the entry "NA" have been transformed to "NaN" in the dataframe.

For example, in the data description file, the column "Alley" should have the following data : "NA" for "No alley access". Instead it was replaced by "NaN" for null.

We will have a look at these columns to make the appropriate modifications.

```python
# Visualize missing values
nulls = houses.null_count().unpivot().filter(pl.col("value") > 0).sort("value", descending=True)
fig = px.bar(
    nulls.to_pandas(), x="variable", y="value",
    color_discrete_sequence=[PASTEL[4]],
    labels={"variable": "Column", "value": "Missing count"},
    title="Missing Values per Column", text_auto=True,
)
fig.update_layout(showlegend=False, width=900, height=400)
fig.show()
```

```python
# Columns where NA means "None/No feature" per data description
na_means_none = [
    "Alley", "BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1",
    "BsmtFinType2", "FireplaceQu", "GarageType", "GarageFinish",
    "GarageQual", "GarageCond", "PoolQC", "Fence", "MiscFeature",
]
houses = houses.with_columns(
    pl.col(c).fill_null("NA") for c in na_means_none if c in houses.columns
)

# LotFrontage: fill with mode
lot_mode = houses["LotFrontage"].drop_nulls().mode().item()
houses = houses.with_columns(pl.col("LotFrontage").fill_null(lot_mode))

# MasVnrType & MasVnrArea: fill with mode
houses = houses.with_columns(
    pl.col("MasVnrType").fill_null(houses["MasVnrType"].drop_nulls().mode().item()),
    pl.col("MasVnrArea").fill_null(houses["MasVnrArea"].drop_nulls().mode().item()),
)

# Electrical: single null — fill with mode
houses = houses.with_columns(
    pl.col("Electrical").fill_null(houses["Electrical"].drop_nulls().mode().item())
)

# GarageYrBlt: fill with YearBuilt (no garage → assume built with the house)
houses = houses.with_columns(
    pl.col("GarageYrBlt").fill_null(pl.col("YearBuilt"))
)

# Verify no nulls remain
remaining = houses.null_count().unpivot().filter(pl.col("value") > 0)
print(f"Remaining nulls: {remaining.height} columns" if remaining.height else "No nulls remain.")
```

### Correlation matrix between columns with numerical values

Now, let's start building a correlation matrix to have a look at which data might have an impact on the sales price.

First, we will store the correlation data into a separate dataset and then we will filter to keep only potentially relevant correlations. Otherwise it would simply be unreadable due to the massive amount of data at our disposal.

```python
# Numeric correlation with SalePrice (sorted)
numeric_cols = [c for c in houses.columns if houses[c].dtype in (pl.Float64, pl.Int64) and c != "Id"]
corr_sale = houses.select(numeric_cols).to_pandas().corr()["SalePrice"].drop("SalePrice").sort_values(ascending=False)

fig = px.bar(
    x=corr_sale.values, y=corr_sale.index,
    orientation="h", color=corr_sale.values,
    color_continuous_scale="RdBu_r", color_continuous_midpoint=0,
    title="Numeric Feature Correlation with SalePrice",
    labels={"x": "Correlation", "y": "Feature"},
)
fig.update_layout(height=800, width=700, yaxis=dict(autorange="reversed"))
fig.show()
```

#### How does it work?

The values are between -1 and +1.

A positive correlation is observed when the values of the two variables being compared increase together.

And a negative correlation when the values of one variable increase while the values of the other variable decrease.

At first glance, some values seem to have a high correlation with the sale price of a house:
- OverallQual has a 0.79 correlation with SalePrice
- GrLivArea has a 0.71 correlation with SalePrice
- GarageCars has a 0.64 correlation with SalePrice

And so on.

Also, some seem to have almost no impact on the price of a house : all those values where skipped because we requested for our correlations to be superior to 0.20 or inferior to -0.20. So all the columns that were not included in the correlation matrix could be considered to have a limited or no positive or negative impact on the price of a house.

```python
# Full correlation heatmap (top features only, above |0.4|)
top_features = corr_sale[corr_sale.abs() >= 0.4].index.tolist() + ["SalePrice"]
corr_top = houses.select(top_features).to_pandas().corr()

fig = px.imshow(
    corr_top, text_auto=".2f", color_continuous_scale="RdBu_r",
    zmin=-1, zmax=1, title="Correlation Matrix (|r| ≥ 0.4 with SalePrice)",
)
fig.update_layout(width=800, height=700)
fig.show()
```

We can consider the columns above 0.40 to have an impact on the sales price of a house. These columns are :
- OverallQual for the overall quality of the house
- GrLivArea for the above ground living area measured in square feet
- GarageCars for the size of the garage measured in car capacity
- GarageArea for the size of the garage measured in square feet
- FullBath for the number of bathrooms above ground
- TotalBsmtSF for the total square feet of the basement area
- 1stFlrSF for the size of the first floor measured in square feet
- TotRmsAbvGrd for the total number of rooms above ground (excluding bathrooms)
- YearBuilt for the year in which the house was built
- YearRemodAdd for remodel date
- MasVnrArea for the masonry veneer area in square feet
- Fireplaces for the number of fireplaces in the house

A few notes:
- GarageCars and GarageArea are pretty similar, they represent the size of the garage but are simply measured differently : one is measured with square feet, the other is measured with the number of cars that could fit in the garage.
- Most of these represent sizes in square feet which could mean that the size does count !
- All of the columns used have numerical data, categorical data was not included in these correlations.

### Correlation with encoded categorical values

Now, to be a bit more thorough, we will transform categorical data using ordinal encoding in order to add them to our correlation analysis.

```python
# Ordinal-encode all string columns for correlation analysis
houses_encoded = houses.drop("Id").clone()
for col_name in houses_encoded.columns:
    if houses_encoded[col_name].dtype == pl.String:
        cats = sorted(houses_encoded[col_name].unique().to_list())
        mapping = {v: i for i, v in enumerate(cats)}
        houses_encoded = houses_encoded.with_columns(
            pl.col(col_name).replace_strict(mapping, return_dtype=pl.Int64)
        )

# Full correlation with SalePrice including encoded categoricals
corr_full = houses_encoded.to_pandas().corr()["SalePrice"].drop("SalePrice").sort_values(ascending=False)

fig = px.bar(
    x=corr_full.values, y=corr_full.index,
    orientation="h", color=corr_full.values,
    color_continuous_scale="RdBu_r", color_continuous_midpoint=0,
    title="All Features Correlation with SalePrice (incl. encoded categoricals)",
    labels={"x": "Correlation", "y": "Feature"},
)
fig.update_layout(height=1000, width=700, yaxis=dict(autorange="reversed"))
fig.show()
```

A few notes :
- The above 0.40 ranking did not change.
- The below or equal to -0.40 ranking did change : some of our categorical data is shown as having a correlation on the sales price of the house.

The columns that are below or equal to -0.40 are, starting from the lowest of all :
- ExterQual for the quality of the material on the exterior of the house
- BsmtQual for the evaluation of the height of the basement
- KitchenQual for the evaluation of the kitchen quality
- GarageFinish for the quality of the interior finish of the garage
- GarageType for the type of the garage and its location compared to the house
- HeatingQC for the heating quality and condition

These columns seem to be linked to quality and the evaluation of quality for certain aspects of a house. I wonder if we could link these elements to the column "OverallQual" (for overall quality) which sits at the top of the correlation heatmap.

### Correlation with the overall quality of a house

```python
corr_quality = houses_encoded.to_pandas().corr()["OverallQual"].drop("OverallQual").sort_values(ascending=False)

fig = px.bar(
    x=corr_quality.values, y=corr_quality.index,
    orientation="h", color=corr_quality.values,
    color_continuous_scale="RdBu_r", color_continuous_midpoint=0,
    title="All Features Correlation with OverallQual",
    labels={"x": "Correlation", "y": "Feature"},
)
fig.update_layout(height=1000, width=700, yaxis=dict(autorange="reversed"))
fig.show()
```

### The Case of the Neighborhood

I have to admit that I'm a bit surprised that the column "Neighborhood" has a correlation of 0.21 with the "SalePrice" column. We would tend to think that the neighborhood would have an influence, positive or negative, on the price. We will test that theory.

```python
# Sales count per neighborhood
neigh_counts = houses["Neighborhood"].value_counts().sort("count", descending=True)
fig = px.bar(
    neigh_counts.to_pandas(), x="Neighborhood", y="count",
    color_discrete_sequence=[PASTEL[0]],
    title="Number of Sales per Neighborhood", text_auto=True,
)
fig.update_layout(width=900, height=400)
fig.show()
```

```python
# Neighborhood impact on price — box plot (much better than strip plot for distributions)
neigh_order = (
    houses.group_by("Neighborhood")
    .agg(pl.col("SalePrice").median().alias("median_price"))
    .sort("median_price", descending=True)
)["Neighborhood"].to_list()

fig = px.box(
    houses.to_pandas(), x="Neighborhood", y="SalePrice",
    color_discrete_sequence=[PASTEL[3]],
    title="SalePrice Distribution by Neighborhood",
    category_orders={"Neighborhood": neigh_order},
)
fig.update_layout(width=1000, height=500)
fig.show()
```

There is a real discrepancy between the highest and lowest neighborhoods in term of mean and median amount of dollars required to purchase a house. This confirms that the neighborhood indeed has an impact on the sales price.

### Category group comparisons

It may be interesting to put some columns in similar categories and compare their correlation with SalePrice.

```python
# Define column groups
groups = {
    "Lot & Sale": [
        "Alley", "Condition1", "Condition2", "LandContour", "LotArea", "LotConfig",
        "LotFrontage", "LotShape", "MSSubClass", "MSZoning", "Neighborhood", "Street",
        "Utilities", "MiscFeature", "MiscVal", "MoSold", "YrSold", "SaleType", "SaleCondition",
    ],
    "Building Size & Features": [
        "GrLivArea", "1stFlrSF", "2ndFlrSF", "LowQualFinSF",
        "CentralAir", "Electrical", "Fireplaces", "FireplaceQu",
        "Functional", "Heating", "HeatingQC",
    ],
    "Building Details": [
        "BldgType", "Exterior1st", "Exterior2nd", "ExterQual", "ExterCond",
        "Foundation", "HouseStyle", "MasVnrType", "MasVnrArea",
        "OverallQual", "OverallCond", "RoofStyle", "RoofMatl", "YearBuilt", "YearRemodAdd",
    ],
    "Interior Spaces": [
        "BsmtCond", "BsmtExposure", "BsmtFinType1", "BsmtFinSF1",
        "BsmtFinType2", "BsmtFinSF2", "BsmtQual", "BsmtUnfSF", "TotalBsmtSF",
        "BedroomAbvGr", "BsmtFullBath", "BsmtHalfBath", "FullBath",
        "HalfBath", "KitchenAbvGr", "KitchenQual", "TotRmsAbvGrd",
    ],
    "Garage & Exterior": [
        "GarageArea", "GarageCars", "GarageCond", "GarageFinish",
        "GarageQual", "GarageType", "GarageYrBlt",
        "EnclosedPorch", "Fence", "LandSlope", "OpenPorchSF",
        "PavedDrive", "PoolArea", "PoolQC", "ScreenPorch", "WoodDeckSF", "3SsnPorch",
    ],
}

# Show per-group correlation heatmaps with SalePrice
for group_name, cols in groups.items():
    valid = [c for c in cols if c in houses_encoded.columns] + ["SalePrice"]
    grp_corr = houses_encoded.select(valid).to_pandas().corr()["SalePrice"].drop("SalePrice").sort_values(ascending=False)
    fig = px.bar(
        x=grp_corr.values, y=grp_corr.index,
        orientation="h", color=grp_corr.values,
        color_continuous_scale="RdBu_r", color_continuous_midpoint=0,
        title=f"{group_name} — Correlation with SalePrice",
        labels={"x": "Correlation", "y": "Feature"},
    )
    fig.update_layout(height=max(300, len(grp_corr) * 25), width=700, yaxis=dict(autorange="reversed"))
    fig.show()
```

### Top feature scatter plots vs SalePrice

```python
# Scatter plots: top 6 correlated features vs SalePrice
top6 = ["OverallQual", "GrLivArea", "GarageCars", "GarageArea", "TotalBsmtSF", "1stFlrSF"]

fig = make_subplots(rows=2, cols=3, subplot_titles=top6)
for i, feat in enumerate(top6):
    r, c = divmod(i, 3)
    fig.add_trace(go.Scatter(
        x=houses[feat].to_list(), y=houses["SalePrice"].to_list(),
        mode="markers", marker=dict(color=PASTEL[i], opacity=0.5, size=4),
        showlegend=False,
    ), row=r + 1, col=c + 1)
    fig.update_xaxes(title_text=feat, row=r + 1, col=c + 1)
fig.update_yaxes(title_text="SalePrice", col=1)
fig.update_layout(title="Top Features vs SalePrice", height=600, width=1000)
fig.show()
```
